## 3. Analisis Risiko dengan LLM (Groq API)

Gunakan Groq API dengan model `llama-3.3-70b-versatile` untuk menganalisis kondisi rantai pasok.

```python
# %pip install groq
# %pip install python-dotenv
```

```python
from groq import Groq
from google.colab import userdata
import os
os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
client = Groq(api_key=os.environ["GROQ_API_KEY"])

prompt = """
Analisis risiko rantai pasok beras berikut:
- Stok turun 20%
- Curah hujan naik 40%
- Harga naik 15%

Berikan:
1. Identifikasi risiko utama
2. Dampak terhadap rantai pasok
3. Rekomendasi mitigasi
"""

response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role": "user", "content": prompt}]
)

ai_analysis = response.choices[0].message.content
print(ai_analysis)
```
