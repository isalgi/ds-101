## 1. Latar Belakang Kasus

Pada 23 Maret 2021, kapal kontainer raksasa **Ever Given** kandas secara melintang di Terusan Suez, Mesir — jalur pelayaran yang dilalui sekitar **12% perdagangan dunia**. Kapal tersangkut selama **6 hari** (23–29 Maret 2021), menyebabkan antrean ratusan kapal dan kerugian ekonomi global yang masif.

Dataset di bawah merangkum dampak harian selama periode penyumbatan (angka bersifat estimasi untuk keperluan pembelajaran, disusun dari berbagai laporan publik).

```python
import pandas as pd

suez = pd.DataFrame({
    'tanggal':                ['2021-03-23', '2021-03-24', '2021-03-25', '2021-03-26', '2021-03-27', '2021-03-28', '2021-03-29'],
    'hari_ke':                [1, 2, 3, 4, 5, 6, 7],
    'kapal_mengantre':        [50, 130, 200, 280, 320, 369, 422],
    'estimasi_nilai_tertahan_usd_miliar': [9.6, 19.2, 28.8, 38.4, 48.0, 57.6, 9.6],
    'kenaikan_harga_minyak_persen':       [1.5, 3.8, 4.6, 6.0, 5.1, 4.0, -2.0],
    'status': ['Kandas', 'Upaya pengerukan', 'Upaya pengerukan',
               'Upaya penarikan', 'Upaya penarikan', 'Upaya penarikan', 'Berhasil dibebaskan']
})

suez
```

---
## 2. Analisis Dampak

Menghitung estimasi total kerugian dan pola eskalasi selama masa penyumbatan.

```python
total_nilai_tertahan = suez.loc[suez['hari_ke'] < 7, 'estimasi_nilai_tertahan_usd_miliar'].sum()
puncak_antrean = suez['kapal_mengantre'].max()
hari_puncak = suez.loc[suez['kapal_mengantre'].idxmax(), 'tanggal']

print(f"Estimasi total nilai perdagangan yang tertahan selama blokade: ~USD {total_nilai_tertahan:.1f} miliar")
print(f"Puncak antrean kapal: {puncak_antrean} kapal (pada {hari_puncak})")
print(f"Lama blokade: {suez['hari_ke'].max()} hari")
```

## 3. Analisis Risiko Global dengan LLM (Groq API)

Gunakan Groq API dengan model `llama-3.3-70b-versatile` untuk menganalisis risiko sistemik dari gangguan satu titik (*single point of failure*) dalam rantai pasok global.

```python
# %pip install groq
```

```python
from groq import Groq
from google.colab import userdata
import os
os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
client = Groq(api_key=os.environ["GROQ_API_KEY"])

data_str = suez.to_string(index=False)

prompt = f"""
Berikut data dampak harian dari kandasnya kapal Ever Given di Terusan Suez (Maret 2021),
sebuah jalur yang dilalui ~12% perdagangan dunia:

{data_str}

Berikan analisis:
1. Mengapa gangguan di satu titik (single point of failure) bisa berdampak sebesar ini secara global
2. Sektor/industri apa saja yang paling terdampak dan mengapa
3. Strategi mitigasi jangka pendek dan jangka panjang yang bisa diambil perusahaan global
"""

response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role": "user", "content": prompt}]
)

analisis_global = response.choices[0].message.content
print(analisis_global)
```
